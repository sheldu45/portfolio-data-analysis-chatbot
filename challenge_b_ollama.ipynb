```python
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), 'src')))
from dataframe_bot.main import DataframeBot
```

Initializes the DataframeBot (the Agent) with only model.

```python
try:
    bot = DataframeBot(model="llama3.1")
except Exception as e:
    print(f"Error initializing DataframeBot: {e}")
    sys.exit(1)
```

Set the Agent's dataframe after its initialization. 

During dataframe loading, the agent will induce informations about each column (header_description). When answering the question, this information will be used by the agent along with the dataframe.

```python
path_data = os.path.join(os.getcwd(), "data", "physical_exam_study.csv")

# Load the DataFrame (and get the header description)
header_description = bot.load_dataframe(path_data)
```

Run the Agent with a question (using previously set dataframe).

```python
question = "What is the mean and standard deviation of the workout intensity for men?"
answer = bot.chat(header_description, question)
print("Answer:", answer)
```
